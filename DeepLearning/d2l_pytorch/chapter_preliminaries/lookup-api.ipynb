## 查找模块中的所有函数和类

为了知道模块中可以调用哪些函数和类，可以调用`dir`函数。
例如，我们可以(**查询随机数生成模块中的所有属性：**)


````python
import torch

print(dir(torch.distributions))
````

通常可以忽略以“`__`”（双下划线）开始和结束的函数，它们是Python中的特殊对象，
或以单个“`_`”（单下划线）开始的函数，它们通常是内部函数。
根据剩余的函数名或属性名，我们可能会猜测这个模块提供了各种生成随机数的方法，
包括从均匀分布（`uniform`）、正态分布（`normal`）和多项分布（`multinomial`）中采样。

## 查找特定函数和类的用法

有关如何使用给定函数或类的更具体说明，可以调用`help`函数。
例如，我们来[**查看张量`ones`函数的用法。**]


````python
help(torch.ones)
````

从文档中，我们可以看到`ones`函数创建一个具有指定形状的新张量，并将所有元素值设置为1。
下面来[**运行一个快速测试**]来确认这一解释：


````python
torch.ones(4)
````

在Jupyter记事本中，我们可以使用`?`指令在另一个浏览器窗口中显示文档。
例如，`list?`指令将创建与`help(list)`指令几乎相同的内容，并在新的浏览器窗口中显示它。
此外，如果我们使用两个问号，如`list??`，将显示实现该函数的Python代码。

## Jupyter 中 `?` 和 `??` 的使用示例

### 单问号 `?` - 查看文档

在 Jupyter Notebook 中，使用 `?` 可以快速查看函数或类的文档说明：

```python
# 查看函数文档
torch.ones?

# 查看类文档
torch.Tensor?

# 查看模块文档
torch.distributions?
```

**效果：** 会弹出一个新窗口，显示该对象的详细文档，类似于 `help()` 的输出。

---

### 双问号 `??` - 查看源代码

使用 `??` 可以查看函数或类的 Python 源代码：

```python
# 查看函数源代码
torch.nn.functional.relu??

# 查看类源代码
torch.nn.Linear??
```

**效果：**
- 如果是 Python 实现的函数，会显示完整源代码
- 如果是 C/C++ 实现的函数（如大多数 PyTorch 核心函数），会显示文档但无法显示源代码

---

### 对比三种文档查看方式

| 方式 | 环境 | 显示内容 | 显示位置 |
|------|------|---------|---------|
| `help(obj)` | 任何Python环境 | 文档说明 | 终端直接输出 |
| `obj?` | Jupyter Notebook | 文档说明 | 弹出窗口 |
| `obj??` | Jupyter Notebook | 文档 + 源代码 | 弹出窗口 |

---

### 实际使用示例

**以下代码在 Jupyter Notebook 中运行，会弹出文档窗口：**

```python
# 示例1：查看 torch.zeros 的文档
torch.zeros?
```

```python
# 示例2：查看 torch.nn.Linear 类的文档
torch.nn.Linear?
```

```python
# 示例3：查看 torch.nn.Linear 的源代码（如果可用）
torch.nn.Linear??

# 注意：大多数 PyTorch 核心函数是 C++ 实现的，?? 只能显示文档
# 但 Python 实现的函数/类可以显示源代码
```

---

### 通配符搜索

`?` 和 `??` 还支持通配符搜索：

```python
# 查找 torch 中以 "zero" 开头的所有函数
torch.*zero*?

# 输出类似：
# torch.zeros
# torch.zeros_like
```

```python
# 查找 torch.nn 中包含 "Layer" 的所有类
torch.nn.*Layer*?

# 输出类似：
# torch.nn.LayerNorm
# (如果有其他包含 Layer 的类也会显示)
```

---

### 小技巧

| 技巧 | 说明 |
|------|------|
| 快捷键 Shift+Tab | 在 Jupyter 中，把光标放在函数名上，按 Shift+Tab 也能弹出文档 |
| Tab 自动补全 | 输入部分函数名后按 Tab，可以补全并显示选项 |
| `?` 和 `??` 只在 Jupyter 中有效 | 普通终端中无法使用，要用 `help()` 代替 |

---

### 一句话总结

**在 Jupyter Notebook 中，用 `obj?` 查文档，用 `obj??` 查源代码，用 `obj*?` 做模糊搜索。**
