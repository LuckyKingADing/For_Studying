# softmax回归详细解答

---

## 一、分类问题

### 1.1 回归与分类的区别

| 问题类型 | 回答的问题 | 典型例子 |
|---------|-----------|---------|
| **回归** | "多少？" | 房价多少？胜场数多少？住院几天？ |
| **分类** | "哪一个？" | 是垃圾邮件吗？图像是猫还是狗？ |

### 1.2 标签表示方法

**方法一：整数编码**

```
类别编号：猫=1, 鸡=2, 狗=3

问题：暗示猫 < 鸡 < 狗的大小关系
      但类别之间没有自然顺序！
```

**方法二：独热编码（推荐）**

```
猫 → [1, 0, 0]
鸡 → [0, 1, 0]
狗 → [0, 0, 1]

优点：
  → 每个类别独立，无顺序暗示
  → 只有一个位置是1，其余是0
```

## 二、网络架构

### 2.1 为什么需要多个输出？

**线性回归 vs softmax回归：**

| 模型 | 输出数量 | 用途 |
|------|---------|------|
| 线性回归 | 1个 | 预测一个数值（房价） |
| softmax回归 | 多个（q个） | 预测q个类别的概率 |

**例子：图像分类（猫、鸡、狗）**

```
输入：一张2×2的灰度图像 → 4个像素值
      x₁, x₂, x₃, x₄

输出：3个类别的分数
      o₁ → 猫的分数
      o₂ → 鸡的分数
      o₃ → 狗的分数

最终选择分数最高的类别作为预测结果
```

---

### 2.2 网络结构图解

```
                    输出层（3个类别）

    x₁ ──┬──── w₁₁ ────────→ o₁ (猫)
         │     w₂₁ ────────→ o₂ (鸡)
         │     w₃₁ ────────→ o₃ (狗)
         │
    x₂ ──┼──── w₁₂ ─────→
         │     w₂₂ ─────→
         │     w₃₂ ─────→
         │
    x₃ ──┼──── w₁₃ ─────→
         │     w₂₃ ─────→
         │     w₃₃ ─────→
         │
    x₄ ──┴──── w₁₄ ─────→
               w₂₄ ─────→
               w₃₄ ─────→

    输入层（4个特征）
```

**关键点：全连接层**

```
每个输出 o_j 都依赖于所有输入 x₁~x₄

特点：
  → 输出层每个神经元与所有输入相连
  → 这就是"全连接层"（fully-connected layer）
  → softmax回归的输出层是全连接层
```

---

### 2.3 仿射变换详解

**什么是仿射变换？**

```
仿射变换 = 线性变换 + 平移

线性变换：W × x（旋转、缩放）
平移：+ b（偏移）

公式：o = W × x + b
```

**展开形式（逐个输出）：**

```
猫的分数：
  o₁ = x₁×w₁₁ + x₂×w₁₂ + x₃×w₁₃ + x₄×w₁₄ + b₁
      = 第1类别的加权和 + 偏置

鸡的分数：
  o₂ = x₁×w₂₁ + x₂×w₂₂ + x₃×w₂₃ + x₄×w₂₄ + b₂
      = 第2类别的加权和 + 偏置

狗的分数：
  o₃ = x₁×w₃₁ + x₂×w₃₂ + x₃×w₃₃ + x₄×w₃₄ + b₃
      = 第3类别的加权和 + 偏置
```

**权重解释：**

```
w_jk 的含义：
  → j：类别索引（第j个输出）
  → k：特征索引（第k个输入）

  → w_jk 表示：第k个特征对第j类别的贡献权重

例子：
  w₁₁ = 第1个像素(x₁)对"猫"分类的重要性
  w₂₁ = 第1个像素(x₁)对"鸡"分类的重要性

如果 w₁₁ > w₂₁：
  → x₁越大，越倾向于分类为"猫"而非"鸡"
```

---

### 2.4 向量形式（矩阵运算）

**向量表示：**

```
输入向量：x = [x₁, x₂, x₃, x₄]ᵀ  （形状：4×1）

权重矩阵：W = [w₁₁, w₁₂, w₁₃, w₁₄     （形状：3×4）
               w₂₁, w₂₂, w₂₃, w₂₄
               w₃₁, w₃₂, w₃₃, w₃₄]

偏置向量：b = [b₁, b₂, b₃]ᵀ  （形状：3×1）

输出向量：o = [o₁, o₂, o₃]ᵀ  （形状：3×1）
```

**矩阵运算过程：**

```
第1步：矩阵乘法
  W × x = [w₁₁×x₁ + w₁₂×x₂ + w₁₃×x₃ + w₁₄×x₄]  → 第1行
          [w₂₁×x₁ + w₂₂×x₂ + w₂₃×x₃ + w₂₄×x₄]  → 第2行
          [w₃₁×x₁ + w₃₂×x₂ + w₃₃×x₃ + w₃₄×x₄]  → 第3行

第2步：加偏置
  W × x + b = [w₁₁×x₁ + ... + w₁₄×x₄ + b₁]  → o₁
              [w₂₁×x₁ + ... + w₂₄×x₄ + b₂]  → o₂
              [w₃₁×x₁ + ... + w₃₄×x₄ + b₃]  → o₃

简洁写法：
  o = W × x + b
```

---

### 2.5 参数统计详解

**计算方法：**

```
权重数量：
  → 每个输出需要d个权重（对应d个输入）
  → q个输出，共需要 q×d 个权重

  本例：d=4（特征），q=3（类别）
  → 权重数 = 3×4 = 12 个

偏置数量：
  → 每个输出需要1个偏置
  → q个输出，共需要 q 个偏置

  本例：偏置数 = 3 个

总参数：
  → 总计 = 权重数 + 偏置数 = q×d + q
  → 本例：12 + 3 = 15 个参数
```

**参数表格：**

| 参数类型 | 数量 | 具体说明 |
|---------|------|---------|
| 权重 w₁₁~w₁₄ | 4个 | 猫分类的权重 |
| 权重 w₂₁~w₂₄ | 4个 | 鸡分类的权重 |
| 权重 w₃₁~w₃₄ | 4个 | 狗分类的权重 |
| 偏置 b₁, b₂, b₃ | 3个 | 每类一个偏置 |
| **总计** | **15个** | 12权重 + 3偏置 |

### 2.6 全连接层的参数开销

**问题：参数量随维度爆炸**

```
一般公式：
  输入维度：d
  输出维度：q
  参数量：O(d × q)

例子：

| 任务 | 输入d | 输出q | 参数量 |
|------|-------|-------|--------|
| 小图像分类 | 4 | 3 | 15 |
| MNIST | 784 | 10 | 7,850 |
| 小RGB图 | 1024 | 10 | 10,250 |
| 大RGB图 | 150528 | 1000 | 150M+ |

问题：
  → 参数量太大（150M）
  → 计算慢，存储大
  → 容易过拟合
```

**解决方案：**

```
论文方法（Zhang et al. 2021）：
  → 将参数开销从 O(d×q) 减少到 O(d×q/n)
  → n是超参数，平衡效率和效果

实践方法：
  → 使用卷积层替代全连接层
  → 先降维再分类
  → 使用更深的网络（减少单层参数）
```

### 2.7 与线性回归的对比

| 对比项 | 线性回归 | softmax回归 |
|--------|---------|------------|
| 输出数量 | 1个 | q个（类别数） |
| 权重矩阵形状 | (1, d) | (q, d) |
| 偏置形状 | (1,) | (q,) |
| 输出含义 | 预测数值 | 各类别的分数 |
| 后处理 | 无 | softmax转换为概率 |
| 参数量 | d+1 | q×d+q |

**关键区别：**

```
线性回归：
  ŷ = wᵀx + b  → 单个数值

softmax回归：
  o = Wx + b    → q个分数
  ŷ = softmax(o) → q个概率

softmax回归本质上是：
  → q个独立的线性回归
  → 加上softmax归一化
```

## 三、softmax运算

### 3.1 线性输出的问题

**问题演示：**

```
假设线性输出：
  o₁ = 2.0
  o₂ = 1.0
  o₃ = 0.1

问题1：总和不为1
  2.0 + 1.0 + 0.1 = 3.1 ≠ 1

问题2：可能为负值
  o₁ = -0.5  ← 负值！概率不能为负
```

### 3.2 softmax解决方案

**公式：**

```
ŷ_j = exp(o_j) / [exp(o₁) + exp(o₂) + exp(o₃)]
```

**通俗理解：**

| 步骤 | 操作 | 作用 |
|------|------|------|
| 第1步 | exp(o_j) 指数化 | 确保非负（指数函数值永远>0） |
| 第2步 | 除以总和 | 确保总和=1（归一化） |

### 3.3 数值计算实例

```
输入：o = [2.0, 1.0, 0.1]

第1步：指数化（确保非负）
  exp(2.0) = 7.389
  exp(1.0) = 2.718
  exp(0.1) = 1.105
  
第2步：计算总和
  sum = 7.389 + 2.718 + 1.105 = 11.212

第3步：归一化（确保总和=1）
  ŷ₁ = 7.389 / 11.212 = 0.659 (65.9%)
  ŷ₂ = 2.718 / 11.212 = 0.242 (24.2%)
  ŷ₃ = 1.105 / 11.212 = 0.099 (9.9%)

验证：
  0.659 + 0.242 + 0.099 = 1.0 ✓
  所有值 ≥ 0 ✓
```

### 3.4 softmax不改变大小顺序

```
排序保持：
  o₁ > o₂ > o₃ (原始输出)
  
经过softmax：
  ŷ₁ > ŷ₂ > ŷ₃ (概率输出)
  
最大值对应关系不变：
  argmax(ŷ) = argmax(o)
```

**实际意义：** 预测时直接找最大的o即可，不一定要先算softmax。

## 四、小批量样本的矢量化

### 4.1 维度说明

| 变量 | 形状 | 含义 |
|------|------|------|
| X | n×d | n个样本，每个d个特征 |
| W | d×q | 权重矩阵 |
| b | q | 偏置向量 |
| O | n×q | 未规范化预测 |
| Ŷ | n×q | 概率输出 |

### 4.2 矢量化计算

```
矩阵形式：
  O = X × W + b
  Ŷ = softmax(O) ← 每行独立归一化

例子（3个样本，2个特征，3个类别）：
  
X = [样本1: [x₁₁, x₁₂]
     样本2: [x₂₁, x₂₂]
     样本3: [x₃₁, x₃₂]]

W = [w₁₁, w₁₂, w₁₃
     w₂₁, w₂₂, w₂₃]

O = X × W + b → 得到3×3的预测矩阵

softmax按行执行：
  第1行归一化 → 样本1的3个类别概率
  第2行归一化 → 样本2的3个类别概率
  第3行归一化 → 样本3的3个类别概率
```

## 五、损失函数

### 5.1 交叉熵损失公式

**定义：**

```
l(y, ŷ) = -Σ y_j × log(ŷ_j)
         = -(y₁×log(ŷ₁) + y₂×log(ŷ₂) + y₃×log(ŷ₃))
```

**独热编码简化：**

```
真实标签：y = [0, 0, 1] (狗)
预测概率：ŷ = [0.1, 0.2, 0.7]

损失计算：
  l = -(0×log(0.1) + 0×log(0.2) + 1×log(0.7))
    = -log(0.7)
    ≈ 0.357

只有真实类别的项起作用！
```

### 5.2 损失值的含义

```
预测准确时：
  ŷ_correct = 1.0 → l = -log(1) = 0 (损失最小)

预测错误时：
  ŷ_correct = 0.01 → l = -log(0.01) ≈ 4.6 (损失很大)
  
规律：
  → 预测概率越接近1，损失越接近0
  → 预测概率越接近0，损失越大
```

| 预测概率 ŷ_correct | 损失值 -log(ŷ) |
|-------------------|---------------|
| 0.99 | 0.01 (很小) |
| 0.8 | 0.22 |
| 0.5 | 0.69 |
| 0.1 | 2.30 (很大) |
| 0.01 | 4.61 (非常大) |

### 5.3 softmax的导数（梯度）

**关键公式：**

```
∂l/∂o_j = ŷ_j - y_j
         = 预测概率 - 真实标签
```

**数值例子：**

```
真实标签：y = [0, 0, 1]
预测概率：ŷ = [0.1, 0.2, 0.7]

梯度计算：
  ∂l/∂o₁ = 0.1 - 0 = 0.1  → 正值，需要减小o₁
  ∂l/∂o₂ = 0.2 - 0 = 0.2  → 正值，需要减小o₂
  ∂l/∂o₃ = 0.7 - 1 = -0.3 → 负值，需要增大o₃

直观理解：
  → 错误类别：预测 > 真实 → 梯度正 → 减小输出
  → 正确类别：预测 < 真实 → 梯度负 → 增大输出
```

**与线性回归对比：**

| 模型 | 梯度形式 | 含义 |
|------|----------|------|
| 线性回归 | ŷ - y | 预测值减真实值 |
| softmax回归 | ŷ_j - y_j | 预测概率减真实标签 |

## 六、信息论基础

### 6.1 熵的概念

**公式：**

```
H(P) = -Σ P(j) × log(P(j))
```

**含义：** 量化分布的不确定性

**例子：**

```
情况1：完全确定
  P = [1, 0, 0]  (只有类别1发生)
  H = -1×log(1) = 0  (无不确定性)

情况2：完全不确定
  P = [1/3, 1/3, 1/3]  (三类等概率)
  H = -3×(1/3)×log(1/3) = log(3) ≈ 1.585
  
结论：分布越均匀，熵越大，越不确定
```

| 分布类型 | 分布值 | 熵 H |
|---------|--------|------|
| 确定分布 | [1, 0, 0] | 0 (最小) |
| 二元均匀 | [0.5, 0.5, 0] | 0.693 |
| 三元均匀 | [1/3, 1/3, 1/3] | 1.099 (最大) |

### 6.2 信息量

**公式：**

```
信息量 = -log(P(j))
```

**含义：** 事件的"惊异程度"

| 事件概率 P | 信息量 -log(P) | 解释 |
|-----------|---------------|------|
| P=1（必然发生） | 0 | 不惊异，意料之中 |
| P=0.5 | 0.693 | 有点惊异 |
| P=0.01（罕见） | 4.6 | 非常惊异 |

### 6.3 交叉熵

**公式：**

```
H(P, Q) = -Σ P(j) × log(Q(j))
```

**含义：** 用分布Q编码来自分布P的数据，平均需要的比特数

**分类任务的两种理解：**

```
理解1：最大化似然
  → 让模型预测分布接近真实分布

理解2：最小化惊异
  → 预测准确时，"惊异"程度低
  → 预测错误时，"惊异"程度高
```

### 练习1：softmax交叉熵的二阶导数

#### 问题1.1：计算二阶导数

**一阶导数（已知）：**

```
∂l/∂o_j = softmax(o)_j - y_j
```

**二阶导数（Hessian矩阵）：**

需要计算 softmax_j 对 o_i 的导数。

设 S = Σ exp(o_k)，则 softmax_j = exp(o_j)/S

**情况1：对角元素（i = j）**

```
∂softmax_j/∂o_j = softmax_j × (1 - softmax_j)

推导步骤：
  第1步：写出softmax定义
    softmax_j = exp(o_j)/S
  
  第2步：使用商法则求导
    ∂/∂o_j [exp(o_j)/S]
    = [exp(o_j)×S - exp(o_j)×exp(o_j)] / S²
    = exp(o_j)×(S - exp(o_j)) / S²
  
  第3步：化简
    = [exp(o_j)/S] × [(S - exp(o_j))/S]
    = softmax_j × (1 - softmax_j)
```

**情况2：非对角元素（i ≠ j）**

```
∂softmax_j/∂o_i = -softmax_j × softmax_i

推导步骤：
  第1步：softmax_j = exp(o_j)/S
  
  第2步：对o_i求导（exp(o_j)是常数）
    = exp(o_j) × ∂(1/S)/∂o_i
    = exp(o_j) × (-exp(o_i)/S²)
  
  第3步：化简
    = -[exp(o_j)/S] × [exp(o_i)/S]
    = -softmax_j × softmax_i
```

**Hessian矩阵结构（3类别例子）：**

| 位置 | (1,1) | (1,2) | (1,3) |
|------|-------|-------|-------|
| 公式 | ŷ₁(1-ŷ₁) | -ŷ₁×ŷ₂ | -ŷ₁×ŷ₃ |

| 位置 | (2,1) | (2,2) | (2,3) |
|------|-------|-------|-------|
| 公式 | -ŷ₂×ŷ₁ | ŷ₂(1-ŷ₂) | -ŷ₂×ŷ₃ |

| 位置 | (3,1) | (3,2) | (3,3) |
|------|-------|-------|-------|
| 公式 | -ŷ₃×ŷ₁ | -ŷ₃×ŷ₂ | ŷ₃(1-ŷ₃) |

#### 问题1.2：与方差的关系

**softmax分布的方差计算：**

softmax输出可以看作伯努利分布的概率参数。

```
对于类别j：
  X_j ∈ {0, 1} (是否是类别j)
  P(X_j = 1) = softmax_j
  
期望：
  E[X_j] = softmax_j
  
方差：
  Var(X_j) = E[X_j²] - E[X_j]²
           = softmax_j - softmax_j²  (伯努利分布性质)
           = softmax_j × (1 - softmax_j)
```

**验证匹配：**

| Hessian对角元素 | softmax方差 |
|----------------|------------|
| ∂²l/∂o_j² = ŷ_j(1-ŷ_j) | Var_j = ŷ_j(1-ŷ_j) |

**结论：相等 ✓**

### 练习2：三等概率编码问题

#### 问题2.1：二进制编码的问题

**设定：**

```
三个类别等概率：P = [1/3, 1/3, 1/3]
```

**熵计算：**

```
H(P) = -Σ P(j)×log(P(j))
     = -3 × (1/3) × log(1/3)
     = log(3)
     ≈ 1.585 纳特
```

**尝试固定长度编码：**

```
3个类别 → 需要多少位？
  → 2位能表示4种状态：00, 01, 10, 11
  → 只用3个：00, 01, 10 (浪费1个)

编码长度：
  → 每个事件固定2位
  → 2比特 ≈ 1.386 纳特
```

**问题分析：**

| 项目 | 数值 |
|------|------|
| 理论最小编码长度（熵） | 1.585 纳特 |
| 实际编码长度 | 1.386 纳特 |
| 差值 | 理论 > 实际？矛盾！ |

**解释：**

实际上固定编码长度 = 2比特 > log₂(3) ≈ 1.585 比特

```
浪费计算：
  实际使用：2比特
  理论最小：log₂(3) ≈ 1.585 比特
  浪费：2 - 1.585 ≈ 0.415 比特/事件
```

#### 问题2.2：更好的编码方法

**方法1：联合编码两个观察**

```
两个独立观察：
  → 组合数：3² = 9 种
  → 每种组合概率：1/9
  
编码9种状态：
  → 需要 ⌈log₂(9)⌉ = 4 位 (能表示16种)
  → 平均每位观察：4/2 = 2 位
```

| 方法 | 每观察位数 |
|------|----------|
| 单独编码 | 2 位 |
| 联合2观察 | 2 位 |

效率相同，但可以扩展。

**方法2：联合编码n个观察**

```
n个独立观察：
  → 组合数：3^n 种
  → 需要位数：⌈n × log₂(3)⌉ 位
  
平均每观察：
  → ⌈n × log₂(3)⌉ / n 位
```

**当n → ∞：**

```
极限分析：
  → ⌈n×log₂(3)⌉/n → log₂(3) ≈ 1.585 位
  
结论：
  → 联合编码足够多的观察
  → 平均编码长度趋近熵的极限
```

### 练习3：RealSoftMax证明

#### 问题3.1：证明 > max(a, b)

**定义：**

```
RealSoftMax(a, b) = log(exp(a) + exp(b))
```

**证明步骤：**

```
第1步：利用指数函数性质
  → exp(a) > 0 (永远为正)
  → exp(b) > 0 (永远为正)

第2步：求和大于单项
  → exp(a) + exp(b) > exp(a)  (因为加了正数exp(b))
  → exp(a) + exp(b) > exp(b)  (因为加了正数exp(a))

第3步：取对数（单调递增函数）
  → log(exp(a) + exp(b)) > log(exp(a)) = a
  → log(exp(a) + exp(b)) > log(exp(b)) = b

第4步：综合结论
  → RealSoftMax(a, b) > a 且 > b
  → 因此 > max(a, b) ✓
```

#### 问题3.2：证明缩放后仍大于max

**命题：**

```
λ⁻¹ × RealSoftMax(λa, λb) > max(a, b)
条件：λ > 0
```

**证明步骤：**

```
第1步：应用问题3.1的结果
  → RealSoftMax(λa, λb) > max(λa, λb)

第2步：缩放性质
  → max(λa, λb) = λ × max(a, b)  (λ>0)

第3步：结合两步
  → RealSoftMax(λa, λb) > λ × max(a, b)

第4步：两边乘以λ⁻¹
  → λ⁻¹ × RealSoftMax(λa, λb) > max(a, b) ✓
```

#### 问题3.3：极限收敛证明

**命题：**

```
当λ → ∞时：
  λ⁻¹ × RealSoftMax(λa, λb) → max(a, b)
```

**证明（设a > b）：**

```
第1步：展开表达式
  λ⁻¹ × RealSoftMax(λa, λb)
  = λ⁻¹ × log(exp(λa) + exp(λb))

第2步：提取主导项
  = λ⁻¹ × log[exp(λa) × (1 + exp(λb - λa))]
  = λ⁻¹ × [λa + log(1 + exp(λ(b-a)))]
  = a + λ⁻¹ × log(1 + exp(λ(b-a)))

第3步：分析极限项
  由于 b - a < 0：
    → λ(b-a) → -∞  (当λ→∞)
    → exp(λ(b-a)) → 0
    → log(1 + exp(λ(b-a))) → log(1) = 0
    → λ⁻¹ × 0 → 0

第4步：极限结果
  → a + 0 = a = max(a, b) ✓
```

**直观理解：**

```
λ的调节作用：
  
λ很小（如λ=1）：
  → softmax很"软"
  → 输出介于max和加权平均之间
  
λ很大（如λ=100）：
  → softmax很"硬"
  → 最大值主导，其他值被压制
  
λ→∞：
  → softmax → 真正的max函数
```

| λ值 | RealSoftMax(2,1)/λ | 值 |
|-----|-------------------|-----|
| 1 | log(e²+e) ≈ 2.31 | > 2 |
| 10 | log(e²⁰+e¹⁰)/10 ≈ 2.000045 | 接近2 |
| 100 | ≈ 2.000000 | ≈ 2 (max) |

#### 问题3.4：soft-min的定义

**类比关系：**

```
soft-max：平滑的max函数
soft-min：平滑的min函数
```

**定义：**

```
softmin(a, b) = -softmax(-a, -b)
              = -log(exp(-a) + exp(-b))
```

**性质：**

```
softmin(a, b) < min(a, b)

证明：
  → softmax(-a, -b) > max(-a, -b)
  → max(-a, -b) = -min(a, b)
  → softmax(-a, -b) > -min(a, b)
  → 两边取负：-softmax(-a, -b) < min(a, b) ✓
```

#### 问题3.5：扩展到多个数

**多个数的soft-max：**

```
softmax(a₁, a₂, ..., aₙ) = log(Σ exp(a_i))
                         = log(exp(a₁) + exp(a₂) + ... + exp(aₙ))
```

**性质：**

| 性质 | 公式 |
|------|------|
| 大于max | softmax(a₁,...,aₙ) > max(a₁,...,aₙ) |
| 极限收敛 | λ⁻¹×softmax(λa₁,...,λaₙ) → max(a₁,...,aₙ) 当λ→∞ |

**多个数的soft-min：**

```
softmin(a₁, a₂, ..., aₙ) = -softmax(-a₁, -a₂, ..., -aₙ)
                          = log(Σ exp(-a_i))
                          = log(exp(-a₁) + exp(-a₂) + ... + exp(-aₙ))
```

**性质：**

| 性质 | 公式 |
|------|------|
| 小于min | softmin(a₁,...,aₙ) < min(a₁,...,aₙ) |
| 极限收敛 | λ⁻¹×softmin(λa₁,...,λaₙ) → min(a₁,...,aₙ) 当λ→∞ |

**数值例子（三个数）：**

```
a = [3, 1, 2]

softmax：
  log(e³ + e¹ + e²) = log(20.09 + 2.72 + 7.39) = log(30.2) ≈ 3.41 > 3 ✓

softmin：
  log(e⁻³ + e⁻¹ + e⁻²) = log(0.05 + 0.37 + 0.14) = log(0.56) ≈ -0.58 < 1 ✓
```
